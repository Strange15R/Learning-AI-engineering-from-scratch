Topic 1: Functions and Modules

```python
# Question 1: Simple Function with Parameters

def calculate_area(length, width):  # The function parameters are declared
    area = length * width           # The calculation process that processes area
    return area

length = 5
width = 4
result = calculate_area(length, width)
print("The area of the rectangle is:", result)
```

```python
# Question 2: Greeting Function with Default Parameter

def greet(name, greeting="Hello"):      # Takes the 2nd input by default in case no greetings input is put manually
    print(greeting + " " + name + "!")  # organizes the output text

greet("Sumaiya")
greet("Rayhan", "Hi")
```

```python
# Question 3: Using Built-in Module

import math        # Imports the math library
x = 25
y = math.sqrt(x)   # Performs the squared root function

print (f"1. The square root of {x} is {y}")


a = 2
b = 3
c = math.pow(a, b)   # Performs the power function (a**b)
print (f"2. {a} raised to the power {b} is {c}")
```

Topic 2: File Handling and OOP Basics

```python
# Question 4: File Handling (Write and Read)



file = open("sample.txt", "w")                 # Create and write to the file
file.write("Welcome to Python Programming!")
file.close()


file = open("sample.txt", "r")                 # Open the file in read mode and print its contents
content = file.read()
print(content)
file.close()
```

```python
# Question 5: OOP Basics (Class and Object)

class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display_info(self):
        print("Student Name:", self.name)
        print("Marks:", self.marks)


student1 = Student("Ayesha", 92)               # Create an object of the Student class
student1.display_info()
```

Using zip() and enumerate() and Lists, Tuples, Sets, and Dictionaries

```python
fruits = ["Apple", "Banana", "Mango"]          # List
colors = ("Red", "Green", "Blue")             # Tuple
numbers = {1, 2, 3, 4}                        # Set

student = {                                   # Dictionary
    "name": "Ayesha",
    "marks": 92
}


for item in enumerate(fruits):                # enumerate() gives index and value
    print(item)


subjects = ["Math", "English", "Science"]     # List
marks = [90, 85, 88]                          # List


for item in zip(subjects, marks):             # zip() combines two lists
    print(item)
```
