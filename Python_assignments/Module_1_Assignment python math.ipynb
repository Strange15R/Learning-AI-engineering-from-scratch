```python
#Q1. Simple Calculator Program

a = 0
b = 0
a = input("Enter your first number: ")
b = input("Enter your second number: ")

a = float(a)
b = float(b)

#Necessary in order to not crash the programm.
if a == 0:
  print("Please enter a non-zero value.")
elif b == 0:
  print("Please enter a non-zero value.")

else:

  #Addition:
  c = a+b
  print("Addition result: ", c)

  #Subtraction:
  c = a-b
  print("Subtraction result: ", c)

  #Multiplication:
  c = a*b
  print("Multiplication result: ", c)

  #Division:
  c = a/b
  print("Division result: ", c)
```

```python
#Q3. Temperature Converter


temp = input("Type the temperature in celcius: ")    #Collect the temperature.
C    = float(temp)

F = (C * 9/5) + 32                                   #Calculte to convert in fahrenheit.
print(f"The temperature in Fahrenheit is: {F}°F")
```
