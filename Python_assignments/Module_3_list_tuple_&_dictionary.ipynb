```python
# Question 1: Enumerate with List

fruits = ["apple", "banana", "cherry", "mango"]

for index, fruit in enumerate(fruits):  #Go through the list, print the index and the items.
    print(index, fruit)
```

```python
# Question 2: Zip with Two Lists

students = ["Rahim", "Karim", "Ayesha"]
scores = [85, 90, 78]

for student, score in zip(students, scores):  #Using the built in zip function, merge and organize both lists.
    print(student, "scored", score)
```

```python
# Question 3: Tuple and Loop

subjects = ("English", "Mathematics", "Physics", "Chemistry", "Biology")

for subject in subjects: #Go through the tuple and print them one by one.
    print(subject)
```

```python
# Question 4: Set Operations

set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

# Common elements
print("Common elements:", set1 & set2)

# Unique elements of set1
print("Unique elements of set1:", set1 - set2)
```
