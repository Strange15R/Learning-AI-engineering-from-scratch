```python
# 1. Using for loop to print numbers from 1 to 10.


for i in range(1, 11):        # Takes the condition

    print(i)                  # Prints it > checkes if the condition is met > goes back to the loop > prints again > does it til the condition is met.
```

```python
# 2. Using for loop to print only the even numbers from 1 to 20.


for i in range(1, 21):     # Iterate through the range of number
    if i % 2 == 0:         # Look if it can be divided in two even parts

        print(i)           # If there is no remainder print the number, do this until the end of the range.
```

```python
# 3. Take input of a string, print uppercased, print the length, print the string reversed.

x = input("Enter a string: ")

print(f"The string in uppercase : {x.upper()}")   #Prints in uppercase
print(f"The length of the string is : {len(x)}")  #Prints the length
print(f"The string in reverse : {x[::-1]}")       #Prints the string in reverse
```

```python
# 4. Taking a string from the user and count how many vowels (a, e, i, o, u) are in the string.

string = input ("Type a word : ")
vowels = "aeiouAEIOU"                        # Taking input and setting up the variables
num = 0

for i in string:       # Iterate the string
  if i in vowels:      # Look for matches
    num = num + 1      # Count the number of success


print (f"The number of vowel is : {num}")    # Print the sum of the times the condition is met.
```
