**Question: Online Shopping Cart System**

An e-commerce website wants to develop a basic shopping cart system using Object-Oriented Programming (OOP) in Python.

```python
class ShoppingCart:
    def __init__(self, customer_name, items=None):
        self.customer_name = customer_name
        self.items = items if items is not None else []


    def add_item(self, item_price):
      if item_price > 0:
        self.items.append(item_price)
        return
      else:
        print("Price must be a positive number.")


    def remove_item(self, item_price):
      if item_price in self.items:
        self.items.remove(item_price)
        print(f"Item {item_price} is removed.")
        return
      else:
        print("Item not found")


    def calculate_total(self):
          return sum(self.items)



    def apply_discount(self):
        total_price = self.calculate_total()
        if total_price >= 3000:
          discounted_price = total_price * 0.90
          return discounted_price
        else:
          print("No discount applied")
          return total_price


    def display_cart(self):
        print(f"...{self.customer_name}'s cart...")

        print(f"Items: ")
        for item in self.items:
            print(f"- {item}")

        print(f"Number of items in the cart: {len(self.items)}")
        print(f"Total: ${self.calculate_total()}")
        print(f"Final amount after discount: ${self.apply_discount()}")


    def clear_cart(self):
      del self.items[:]
      print("Cart is cleared")
      return


    def __add__(self, other_cart):
      combined_cart = f"{self.customer_name} & {other_cart.customer_name}"
      combined_items = self.items + other_cart.items
      return ShoppingCart(combined_cart, combined_items)
```

```python
cart1 = ShoppingCart("Raihan")
cart2 = ShoppingCart("Sumaiya")

# Add and remove items from both carts
cart1.add_item(1500)
cart1.add_item(2000)
cart1.add_item(-50)        # invalid-price branch
cart1.add_item(5000)
cart1.add_item(3000)
cart1.add_item(1000)
cart1.remove_item(2000)

cart2.add_item(1200)
cart2.add_item(1500)
cart2.add_item(6000)      # discount branch
cart2.add_item(1800)
cart2.remove_item(999)     # the "not found" branch


# Display individual cart details
cart1.display_cart()
cart2.display_cart()

# Combine both carts
combined_cart = cart1 + cart2

# Display the combined cart details
combined_cart.display_cart()


# Clearing the carts
cart1.clear_cart()
cart2.clear_cart()
combined_cart.clear_cart()
```
